```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

date_columns = ['Job Start', 'Job End']                                                                                                                                                                                       
date_format = '%Y-%m-%dT%H:%M:%S'
                       
df= pd.read_csv('log.txt', na_values='-',parse_dates=date_columns, date_format=date_format)

df['YearMonth'] = df['Job Start'].dt.to_period('M')
df['Elapsed (h)'] = df['Elapsed (h)'].replace('-', np.nan)

df= df.dropna()
print (len(df))
df = df[df['Req Mem (GB)']!= 0]
print (len(df))
#cpu_jobs = cpu_jobs.dropna()
```

```python
plt.rcParams['font.weight'] = 'bold'
plt.rcParams['axes.labelweight'] = 'bold'
plt.rcParams['axes.titleweight'] = 'bold'
plt.rcParams['lines.linewidth'] = 1.5
plt.rcParams['axes.linewidth'] = 1.5
plt.rcParams['xtick.major.width'] = 1.5
plt.rcParams['ytick.major.width'] = 1.5
```

```python
df['Day of Week'] = df['Job Start'].dt.dayofweek
```

```python
cpu_jobs = df
```

```python
cpu_jobs["Unused Mem (GB)"] = ( 
    cpu_jobs["Req Mem (GB)"] - cpu_jobs["Used Mem(GB)"]
)   
cpu_jobs["Unused Mem (%)"] = ( 
    cpu_jobs["Unused Mem (GB)"] / cpu_jobs["Req Mem (GB)"] * 100.0   )

cpu_jobs["Used Mem (%)"] = ( 
    cpu_jobs["Used Mem(GB)"] / cpu_jobs["Req Mem (GB)"] * 100.0   )
```

```python
cpu_jobs.describe()
```

```python
cpu_jobs.replace(-np.inf, np.nan, inplace=True)
cpu_jobs = cpu_jobs.dropna()
```

```python
jobs_per_month = cpu_jobs.groupby('YearMonth').size().reset_index(name='Count')
jobs_per_month.set_index('YearMonth', inplace=True)
```

```python
# Select all jobs where 'Unused Mem (GB)' is negative
negative_unused_mem_jobs = cpu_jobs[cpu_jobs['Used Mem (%)'] > 100.0]
#negative_unused_mem_jobs
```

```python
negative_jobs_per_month = negative_unused_mem_jobs.groupby('YearMonth').size().reset_index(name='Count')
```

```python
grouped_stats = cpu_jobs.groupby('YearMonth')['Used Mem (%)'].agg(
    Min='min',
    Max='max',
    Mean='mean',
    Std= 'std',
    Q1=lambda x: x.quantile(0.25),
    Q3=lambda x: x.quantile(0.75)
).reset_index()

# Print the grouped statistics (optional)
print(grouped_stats)

```

```python
plt.figure(figsize=(12, 6))

# Plot Max, Mean, Q1, and Q3
#plt.plot(grouped_stats['YearMonth'].dt.to_timestamp(), grouped_stats['Max'], label='Max', marker='o')
plt.plot(grouped_stats['YearMonth'].dt.to_timestamp(), grouped_stats['Mean'], label='Mean', marker='o', color = 'blue')
#plt.plot(grouped_stats['YearMonth'].dt.to_timestamp(), grouped_stats['Q1'], label='Q1', marker='o')
#plt.plot(grouped_stats['YearMonth'].dt.to_timestamp(), grouped_stats['Q3'], label='Q3', marker='o')

plt.fill_between(grouped_stats['YearMonth'],
                 grouped_stats['Mean'] - grouped_stats['Mean'],
                 grouped_stats['Mean'] + grouped_stats['Std'],
                 color='blue', alpha=0.2, label='Std Dev')

# Adding titles and labels
#plt.title('Max, Mean, Q1, and Q3 Over Time')
#plt.xlabel('YearMonth')
#plt.ylabel('Values')
plt.xlabel('Time')
plt.ylabel('Used Mem (%) \n (Used Mem - Req Mem)/Req Mem')
plt.xticks(rotation=45)
plt.xlim(pd.to_datetime('2023-03'), pd.to_datetime('2024-08-15'))

# Adding a legend
plt.legend()

# Show the plot
plt.show()
```
