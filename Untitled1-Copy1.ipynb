```python
import pandas as pd
import numpy as np


date_columns = ['Job Start', 'Job End']                                                                                                                                                                                       
date_format = '%Y-%m-%dT%H:%M:%S'
                       
                  
df= pd.read_csv('log_casper.txt', na_values='-',parse_dates=date_columns, date_format=date_format)

df['YearMonth'] = df['Job Start'].dt.to_period('M')
df['Elapsed (h)'] = df['Elapsed (h)'].replace('-', np.nan)
df= df.dropna()
```

```python
df['Day of Week'] = df['Job Start'].dt.dayofweek
```

```python
cpu_jobs = df
```

```python
cpu_jobs["Unused Mem (GB)"] = ( 
    cpu_jobs["Req Mem (GB)"] - cpu_jobs["Used Mem(GB)"]
)   
cpu_jobs["Unused Mem (%)"] = ( 
    cpu_jobs["Unused Mem (GB)"] / cpu_jobs["Req Mem (GB)"] * 100.0   )

cpu_jobs["Used Mem (%)"] = ( 
    cpu_jobs["Used Mem(GB)"] / cpu_jobs["Req Mem (GB)"] * 100.0   )
```

```python
cpu_jobs.describe()
```

```python
cpu_jobs.replace(-np.inf, np.nan, inplace=True)
cpu_jobs = cpu_jobs.dropna()
```

```python
# Select all jobs where 'Unused Mem (GB)' is negative
negative_unused_mem_jobs = cpu_jobs[cpu_jobs['Unused Mem (%)'] < -100.01]

# Display the filtered DataFrame
print(negative_unused_mem_jobs)
```

```python
negative_unused_mem_jobs
```

```python
negative_jobs_per_month = negative_unused_mem_jobs.groupby('YearMonth').size().reset_index(name='Count')
```

```python
negative_jobs_per_month
```

```python
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter


def thousands_formatter(x, pos):
    return f'{int(x):,}'

plt.figure(figsize=(10, 6))
plt.plot(negative_jobs_per_month['YearMonth'].dt.to_timestamp(), negative_jobs_per_month['Count'], marker='o', linestyle='-', color='b')
plt.xlabel('Year-Month')
plt.ylabel('Count')
plt.title('Number of Casper Jobs with >100% Memory Usage')
plt.xticks(rotation=45)
plt.grid(True)
plt.gca().yaxis.set_major_formatter(FuncFormatter(thousands_formatter))

plt.tight_layout()
plt.show()
```
