```python
import pandas as pd
import numpy as np


date_columns = ['Job Start', 'Job End']                                                                                                                                                                                       
date_format = '%Y-%m-%dT%H:%M:%S'
                       
                  
df= pd.read_csv('log.txt', na_values='-',parse_dates=date_columns, date_format=date_format)

df['YearMonth'] = df['Job Start'].dt.to_period('M')
df['Elapsed (h)'] = df['Elapsed (h)'].replace('-', np.nan)
```

```python
data_types = {          
    "Nodes": float,                                                                                                                                                                                                    
    "Elapsed (h)": float,
}                       
df = df.astype(data_types)
```

```python
cpu_jobs = df[df['Queue'].isin(['cpu', 'cpudev'])]
```

```python
cpu_jobs['core-hours'] = np.where(cpu_jobs['Queue'] == 'cpu', 
                                  cpu_jobs['Nodes'] * cpu_jobs['Elapsed (h)'] * 128, 
                                  cpu_jobs['NCPUs'] * cpu_jobs['Elapsed (h)'])
```

```python
grouped_df = cpu_jobs.groupby('User')['core-hours'].sum().reset_index()

sorted_df = grouped_df.sort_values(by='core-hours', ascending=False).reset_index(drop=True)

```

```python
# Group by 'User' and 'YearMonth' and sum 'core-hours'
grouped_df = cpu_jobs.groupby(['User', 'YearMonth'])['core-hours'].sum().reset_index()
```

```python
sorted_df = grouped_df.sort_values(by='core-hours', ascending=False).reset_index(drop=True)
```

```python
sorted_df.groupby('YearMonth').max()
```

```python
# Get the top 10 users per month
top_users_per_month = cpu_jobs.groupby('YearMonth').apply(lambda x: x.nlargest(10, 'core-hours')).reset_index(drop=True)
```

```python
import matplotlib.pyplot as plt
# Plot the results
plt.figure(figsize=(14, 8))
for user, group in top_users_per_month.groupby('User'):
    print (user, group['YearMonth'])
    plt.plot(group['YearMonth'], group['core-hours'], marker='o', linestyle='-', label=user)

plt.xlabel('Time (Year-Month)')
plt.ylabel('Core-hours')
plt.title('Top 10 Users per Month by Core-hours')
plt.legend(title='User', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Group by 'YearMonth' and 'User' and sum 'core-hours'
grouped_df = cpu_jobs.groupby(['YearMonth', 'User'])['core-hours'].sum().reset_index()

# Plot the results
plt.figure(figsize=(14, 8))
for user, group in grouped_df.groupby('User'):
    plt.plot(group['YearMonth'], group['core-hours'], marker='o', linestyle='-', label=user)

plt.xlabel('Time (Year-Month)')
plt.ylabel('Core-hours')
plt.title('Sum of Core-hours per User per Month')
plt.legend(title='User', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
